### Seção 1: Criação de funções

```python
# Definindo uma função
def primeiraFunc():
    print("Olá, mundo!")
```

```python
primeiraFunc()
```

```python
# Função para somar números
def addNum(firstNum, secondNum):
    print('Esse é o primeiro númeor: '+str(firstNum))
    print('Esse é o segundo número: '+str(secondNum))
    print('A soma dos números é: ', firstNum+secondNum)
```

```python
addNum(5, 10)
```

```python
# Criando uma função para cálculo de média de 2 valores
def mediaX(x, y):
    return (x+y)/2
```

```python
# testando a função
mediaX(10, 2)
```

### Seção 2: Variáveis locais e globais

```python
varGlobal=10 # esse é um exemplo de variável global
def multiplicação(num1, num2):
    varLocal=num1*num2 # esse é um exemplo de variável local
    print(varLocal)
```

```python
# Mostrando a variável local por meio da função "multiplicação"
multiplicação(5, 10)
```

```python
# Mostrando a variável global
print(varGlobal)
```

```python
# Tentando acessar a variável local diretamentem, olhe o erro.
# Ou seja, é impossível acessar uma variavel local fora da função onde ela foi criada.
print(varLocal)
```
