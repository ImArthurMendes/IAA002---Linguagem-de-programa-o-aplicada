## Parte 2 - Strings
### Seção 1: Criando uma string
#### Para criar uma string em Python você pode usar aspas simples ou duplas. Por exemplo:

```python
# Uma única palavra
'Olá'
```

```python
# Uma frase
'Olá, tudo certo?'
```

```python
# Exemplo usando aspas duplas
"Olá, tudo certo?"
```

```python
# Posso combinar simples e duplas
'Olá, tudo certo? "Sim, tudo ótimo!"'
```

#### Imprimindo uma string

```python
print('Olá, tudo certo?')
```

```python
# Com quebra de linha
print('Olá, tudo certo?\nSim, tudo ótimo!')
```

#### Seção 2: Propriedades das strings

```python
# Criando uma string
frase='Póx IAA - UFPR'
print(frase)
```

```python
# Alterando caractereres de uma string
frase=frase.replace('x', 's')
print(frase)
```

```python
# Concatenando strings
frase=frase+' - Por uma oportunidade melhor na carreira, se DEUS quiser!'
print(frase)                                                 
```

#### Seção 3: Exemplos de funções built-in de strings

```python
# Upper Case (deixar tudo em maiúsculo)
frase.upper()
```

```python
# Lower Case (deixar tudo em minúsculo)
frase.lower()
```

```python
# Dividir uma string por espaços em branco (padrão). Ou posso separar por algum caractere específico, como vírgula, por exemplo, basta usar o caractere como argumento do método split.
frase.split()
```

```python
# Deixa apenas a primeira letra da string em maiúsculo
frase.capitalize()
```

### Parte 4 - Dicionários
#### Seção 1: Criação de dicionários

```python
# Dicionários - uso de chaves e separação de chaves e valores por dois pontos (:)
estudantesDicionario={'Ademico': 18, 'Maneco': 28, 'Frederico': 23, 'Jofrê': 25}
print(estudantesDicionario)
```

```python
# Uso da função type para verificar o tipo de dado de uma variável - dict é a abreviação de dicionário
type(estudantesDicionario)
```

```python
# Diferença para listas e dicionários: listas são ordenadas e indexadas por números, enquanto dicionários são desordenados e indexados por chaves. Em uma lista, os itens são acessados por sua posição (índice), enquanto em um dicionário, os itens são acessados por suas chaves.
estudantesLista=['Ademico', 18, 'Maneco', 28, 'Frederico', 23, 'Jofrê', 25]
print(estudantesLista)
```

```python
type(estudantesLista)
```

```python
# Selecionando um elemento de um dicionário usando a chave
estudantesDicionario["Jofrê"]
```

#### Seção 2: Funções e métodos de dicionários

```python
# Excluindo todos os elementos deum dicionário - método clear()
estudantesDicionario.clear()
print(estudantesDicionario)
```

```python
# Forma usada para deletar variáveis que não estão sedo utilizadas mais
del estudantesDicionario
del estudantesLista
```

```python
# Criando novo dicionário
notaAlunos={'Ademico': 8.5, 'Maneco': 7.0, 'Frederico': 9.0, 'Jofrê': 6.5}
print(notaAlunos)
```

```python
len(notaAlunos)
```

```python
max(notaAlunos)
```

```python
min(notaAlunos)
```

```python
# Retorna apenas as chaves do dicionário
notaAlunos.keys()
```

```python
# Retorna apenas os valores do dicionário
notaAlunos.values()
```

```python
max(notaAlunos.values())
```

```python
min(notaAlunos.values())
```

```python
# Método items retorno os valores em formato de lista
notaAlunos.items()
```

#### Seção 3: Criando dicionários aninhados

```python
# Exemplo de dicionário aninhado
dicionarioAninhado={'aluno1': {'aluno2Aninhado':{'aluno3Aninhado':'Dicionário Aninhado em Python'}}}
```

```python
print(dicionarioAninhado)
```

#### Nesse caso, recebendo o valor das chaves em uma variável "dicionarioAninhado". Note que para acessar o valor da terceira chave ("aluno3Aninhado") é necessário acessar os valores a partir de "aluno1" e "aluno2Aninhado". Ou seja, para acessar a camada da terceira chave, eu tenho que acessar as camadas superiores (chave 1 e 2).

```python
# Exemplo do erro
dicionarioAninhado['aluno3Aninhado']
```

```python
# Exemplo acessando a camada corretamente
dicionarioAninhado['aluno1']['aluno2Aninhado']['aluno3Aninhado']
```

## Parte 5 - Tuplas
### Seção 1: Criação de tuplas

```python
# Estrutura de uma tupla
tupla=('Ademico', 25, 'Maneco')
print(tupla)
```

```python
# Usantos a função type para verificar o tipo de dado da tupla (tuple = tupla em inglês)
type(tupla)
```

```python
# Tuplas não suportam o método append, ou seja, não é possível adicionar um item a uma tupla depois de criada. Veja mensagem de erro:
tupla.append('Frederico')
```

```python
# Tuplas nãi suportam o método remove ou delete, ou seja, não é possível remover um item de uma tupla depois de criada. Veja mensagem de erro:
del tupla['frederico']
```

```python
tupla.remove('Frederico')
```

#### Seção 2: Funções e métodos de tuplas

```python
# Verificando o comprimento da tupla usando a função len()
len(tupla)
```

```python
# Tuplas não suportam atribuições de itens
tupla[2]='Galileu'
```

```python
# Para deixar a Tupla maleável, eu posso transforma-la em uma lista, usando a função list()
tupla=list(tupla)
```

```python
type(tupla)
```

```python
print(tupla)
```

```python
tupla.remove('Ademico')
print(tupla)
```

```python
len(tupla)
```

```python
type(tupla)
```

```python
tupla=tuple(tupla)
```

```python
type(tupla)
```
