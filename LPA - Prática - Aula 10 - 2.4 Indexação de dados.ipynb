## LPA - Prática - Aula 10 - 2.4 Indexação de dados
### Parte 1: Indexação de strings

#### Lembrando: Indexação é quando queremos acessar algum elemento dentro da nossa string, lista, dicionário ou tupla

```python
# Criando uma String
frase = "Ademico pé de pico, Maneco pé de macaco, Frederico pita fumo e Galileu fuma tabaco."
```

```python
print(frase)
```

```python
# Primeiro elemento da string
frase[0]
```

```python
frase[1]
```

##### Podemos usar '1:' para executar um slicing que faz a leitura de tudo até um ponto designado. Por exemplo:

```python
# Retornar todos os elementos da string
frase[:]
```

```python
# Retornar todos os elementos da string, após o primeiro elemento (0)
frase[1:]
```

```python
# Retornar os elemento de uma string até um ponto específico
frase[0:7]
```

```python
# Retornar os elementos da string de tras para frente
frase[:20]
```

##### Nós também podemos usar a notação de índice e fatiar a string em pedaços específicos (o padrão é 1). Por exemplo, podemos usar dois pontos duas vezes em uma linha e, em seguida, um número que especifica a frequência para retornar elementos. Por exemplo

```python
frase[::1]
```

```python
frase[::2]
```

```python
# Com indexação negativa
frase[::-1]
```

#### Parte 2: Indexação de listas

```python
# Criando uma lista
capitaisBrasileiras=['Curitiba', 'Florianópolis', 'Porto Alegre', 'Boa Vista']
print(capitaisBrasileiras)
```

```python
# Quantidade de elementos
len(capitaisBrasileiras)
```

```python
# Tipo de elementos
type(capitaisBrasileiras)
```

```python
# Para consultar um elemento
capitaisBrasileiras[3]
```

#### Podemos usar '1:' para executar um slicing que faz a leitura de tudo até um ponrto designado. Por exemplo:

```python
# A string original permanece inalterasa
capitaisBrasileiras[1:]
```

```python
capitaisBrasileiras
```

```python
# Todos os elementos da lista
capitaisBrasileiras[:]
```

#### Nós também podemos usar a notação de índice e fatiar a lista em pedaços específicos (o padrão é 1)

```python
# O primeiro e o terceito elemento da lista (conta de dois em dois)
capitaisBrasileiras[::2]
```

```python
capitaisBrasileiras[::-2]
```

### Parte 3: Indexação de dicionários

```python
# Criando um dicionário de listas
dicionarioCapitais={'brasilSul': ['Porto Alegre', 'Florianópolis', 'Curitiba'], 'brasilSuldeste': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']}
```

```python
# consultando indices do dicionário
print(dicionarioCapitais['brasilSul'])
print(dicionarioCapitais['brasilSuldeste'])

# consultando indices individuais do dicionario
print(dicionarioCapitais['brasilSul'][2])
print(dicionarioCapitais['brasilSuldeste'][1])

# aplicando o slicing no dicionário
print(dicionarioCapitais['brasilSul'[:]])
```

### Parte 4: Indexação de Tuplas
#### Igual à indexação de listas, com a diferença que as tuplas são imutáveis (não é possível adicionar ou remover valores)

```python
tuplasCapitais=('Curitiba', 'Florianópolis', 'Porto Alegre')
tuplasEstados=('Paraná', 'Santa Catarina', 'Rio Grande do Sul')
```

```python
type(tuplasCapitais)
```

```python
len(tuplasCapitais)
```

```python
tuplaBrasil=tuplasEstados+tuplasCapitais
print(tuplaBrasil)
```

```python
len(tuplaBrasil)
```

```python
tuplaBrasil[4]
```

```python
tuplaBrasil[:3]
```

```python
tuplaBrasil[3:]
```

```python
tuplaBrasil[::2]
```
