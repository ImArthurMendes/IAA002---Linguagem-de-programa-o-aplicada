# LPA - Prática - Aula 14 - 2.8 Expressões Lambda
### Seção 1: Expressões Lambda

```python
# Definindo uma função normal (3 linhas de código)
def potencia(num):
    result=num**2
    return result
```

```python
# Testando a função criada
potencia(5)
```

```python
# Definindo uma função (2 linhas de código)
def potencia(num):
    return num**2
```

```python
# Testando a mesma função
potencia(5)
```

```python
# Definindo uma função (1 linha de código)
def potencia(num): return num**2
```

```python
# Testando a mesma função
potencia(5)
```

```python
# Criando uma função a partir de uma expressão Lambda (1 linha de código)
potencia=lambda num: num**2 
```

```python
# Retorna o mesmo resultado, porém, a criação da função foi simplificada
potencia(5)
```

```python
# Lembrete: operadores de comparação retornam sempre um valor booleano (true ou false)
Par=lambda x: x%2==0 #== é o operador de comparação de igualdade e retorna true se o número for par e false se for ímpar
```

```python
Par(3)
```

```python
Par(6)
```
