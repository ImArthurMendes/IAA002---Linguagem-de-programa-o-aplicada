### Seção 1: Condicional if

```python
# Criando uma condicional if
idade=18
if idade >= 18:
    print('Pode entrar, é maior de idade!')
```

```python
# Se a condição foi falsa, o sistema simplesmente não gera nada. Ou seja, a mensagem do print não será impressa!
if idade < 18:
    print('Não pode entrar, é menor de idade!')
```

### Seção 2: Condicional if-else (se-caso contrário)

```python
# Criando uma condicional if-else para mostrar como o sistema se comporta quando a condição é falsa.
if idade < 18: #se
    print('Maior de idade, pode entrar!')
else: #caso contrário
    print('Menor de idade, não pode entrar!')
```

```python
if idade == 18:
    print('Tem 18 anos')
```

### Seção 3: Condicional aninhados

```python
# Condicional if aninhada. O código só irá para o else, se os dois if's forem falsos. Se apenas um if for verdadeiro,
# o sistema não retornará nenhuma informação presente no print e no else.
nome='Arthur'
if idade >= 18:
    if nome == 'Ademico':
        print(nome, 'pode entrar, você é maior de idade!')
else:
    print(nome, 'você não poderá entrar, pois é menor de idade!')
```

### Seção 4: Condicional elif

```python
nome='Ademico'

if nome == 'Ademico':
    print(nome, 'pé de pico')
elif nome == 'Maneco':
    print(nome, 'pé de macaco')
elif nome == 'Frederico':
    print(nome, 'pita fumo')
else:
    print('Galileu fede tabaco')
```

### Seção 5: Operadores Lógicos

```python
idade=17
nome='Ademico'
if idade >= 18 and nome == 'Ademico':
    print(nome, 'pode entrar, você é maior de idade!')
elif idade >= 18 or nome == 'Ademico':
    print(nome, 'pode entrar, pois é maior de idade ou tem o nome certo!')
elif idade < 18 and nome != 'frederico':
    print(nome, 'você não poderá entrar, pois é menor de idade e tem o nome errado!')
else:
    print(nome, 'você não poderá entrar, pois é menor de idade!')
```
